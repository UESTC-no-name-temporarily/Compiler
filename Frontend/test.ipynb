```python
# Importing required libraries
import os
import subprocess

# Define the directories and file paths
test_cases_dir = "./test_cases"
test_cases_output_dir = "../test_cases_output"

# Ensure the test_cases_output directory exists
os.makedirs(test_cases_output_dir, exist_ok=True)

# Main function
def main():
    # Attempt to open the test cases directory
    try:
        files = sorted(os.listdir(test_cases_dir))  # Sort files for consistent order
    except FileNotFoundError:
        print("Failed to open ./test_cases directory")
        return 1

    # Initialize counters
    passed_count = 0
    total_count = 0

    # Iterate over files in the directory
    for file_name in files:
        # Skip directories
        if os.path.isdir(os.path.join(test_cases_dir, file_name)):
            continue

        # Check for .sy files
        if not file_name.endswith(".sy"):
            continue

        total_count += 1  # Increment the total test case count

        # Extract the base file name (without extension)
        base_name = file_name[:-3]

        # Construct the command
        cmd = f"./build/main < {os.path.join(test_cases_dir, file_name)} > {os.path.join(test_cases_output_dir, base_name)}.out"
        print(f"Testing {file_name}")

        # Execute the command
        try:
            subprocess.run(cmd, shell=True, check=True)
            print("OK")
            passed_count += 1  # Increment the passed test case count
        except subprocess.CalledProcessError:
            print(f"Failed in {file_name}")

    # Print the final results
    print(f"Test done! {passed_count}/{total_count} test cases passed.")
    return 0

# Run the main function
if __name__ == "__main__":
    main()
```
